```python
from tensorflow.keras.applications.vgg16 import VGG16
```

```python
model = VGG16()
```

```python
model.summary()
```

```python
for i in range(len(model.layers)):
    if 'conv' not in str(model.layers[i]):
        continue

    else:
        print(f"Conv_layer{i}: {model.layers[i]}")
```
