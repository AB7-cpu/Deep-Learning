```python
df = pd.DataFrame([[8,8,1], [7,9,1], [6,10,0], [5,5,0]], columns=['cgpa', 'profile_score', 'placed'])
```

```python
def initialize_parameters(layer_dimension):
    np.random.seed(3)
    parameters = {}
    L = len(layer_dimension)

    for i in range(1, L):
        parameters['W' + str(i)] = np.ones((layer_dimension[i-1], layer_dimension[i]))*0.1
        parameters['b' + str(i)] = np.zeros((layer_dimension[i], 1))

    return parameters
```

```python
def sigmoid(Z):

    A = 1/(1+(np.exp(-Z)))

    return A
```

```python
def linear_forward(X, W, b):

    Z = np.dot(W.T, X) + b
    A = sigmoid(Z)

    return A
```

```python
def L_layer_forward(X, parameters):

    A = X
    L = len(parameters) // 2

    for l in range(1, L+1):

        A_prev = A
        W = parameters['W' + str(l)]
        b = parameters['b' + str(l)]

        # print("X" + str(l-1) + ": ", A_prev)
        # print("W"+str(l) + ": ", W)
        # print("b"+str(l) + ": ", b)
        # print("--"*20)

        A = linear_forward(A_prev, W, b)
        # print("A"+str(l)+": ", A)
        # print("**"*20)

    return A,A_prev
```

```python
def update_parameters(parameters, y, y_hat, A1, X, a):
    parameters['W2'][0][0] = parameters['W2'][0][0] + (a * (y-y_hat) * A1[0][0])
    parameters['W2'][1][0] = parameters['W2'][1][0] + (a * (y-y_hat) * A1[1][0])
    parameters['b2'][0][0] = parameters['W2'][1][0] + (a * (y-y_hat))

    parameters['W1'][0][0] = parameters['W1'][0][0] + ( a * (y-y_hat) * parameters['W2'][0][0] * (A1[0][0]*(1-A1[0][0])) * X[0][0] )
    parameters['W1'][0][1] = parameters['W1'][0][1] + ( a * (y-y_hat) * parameters['W2'][0][0] * (A1[0][0]*(1-A1[0][0])) * X[1][0] )
    parameters['b1'][0][0] = parameters['W1'][0][0] + ( a * (y-y_hat) * parameters['W2'][0][0] * (A1[0][0]*(1-A1[0][0])) )

    parameters['W1'][1][0] = parameters['W1'][1][0] + ( a * (y-y_hat) * parameters['W2'][1][0] * (A1[1][0]*(1-A1[1][0])) * X[0][0] )
    parameters['W1'][1][1] = parameters['W1'][1][1] + ( a * (y-y_hat) * parameters['W2'][1][0] * (A1[1][0]*(1-A1[1][0])) * X[1][0] )
    parameters['b1'][1][0] = parameters['W1'][1][0] + ( a * (y-y_hat) * parameters['W2'][1][0] * (A1[1][0]*(1-A1[1][0])) )
```

```python
#epochs implementation:

parameters = initialize_parameters([2,2,1])
epochs = 50

for i in range(epochs):

    Loss = []

    for j in range(df.shape[0]):
        X = df[['cgpa', 'profile_score']].values[j].reshape(2,1)
        y = df[['placed']].values[j][0]

        #parameters initialization

        y_hat, A1 = L_layer_forward(X, parameters)
        y_hat = y_hat[0][0]

        update_parameters(parameters, y, y_hat, A1, X, a=0.001)

        Loss.append(-y*np.log(y_hat) - (1-y)*np.log(1-y_hat))

    print('Epoch - ', i+1, 'Loss - ', np.array(Loss).mean())

parameters
```
