```python
from tensorflow.keras.preprocessing.text import Tokenizer
```

```python
tokenizer = Tokenizer(oov_token='hmm')
```

```python
vocabulary= ['go india',
'india india',
'hip hip hurray',
'jeetega bhai jeetega india jeetega',
'bharat mata ki jai',
'kohli kohli',
'sachin sachin',
'dhoni dhoni',
'modi ji ki jai',
'inquilab zindabad'
]
```

```python
tokenizer.fit_on_texts(vocabulary)
```

```python
tokenizer.word_index
```

```python
sequences = tokenizer.texts_to_sequences(vocabulary)
```

```python
sequences
```

```python
from tensorflow.keras.utils import pad_sequences
```

```python
sequences = pad_sequences(sequences, padding='post')
```

```python
sequences
```

```python
from tensorflow.keras.datasets import imdb
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense, SimpleRNN, Embedding, Flatten
```

```python
(X_train, y_train), (X_test, y_test) = imdb.load_data()
```

```python
X_train[:5]
```

```python
y_train[:5]
```

```python
X_train = pad_sequences(X_train, padding='post', maxlen=200)
X_test = pad_sequences(X_test, padding='post', maxlen=200)
```

```python
X_train.shape
```

```python
model = Sequential()
```

```python
model.add(SimpleRNN(32, input_shape=(200,1), return_sequences=False, activation='relu'))
model.add(Dense(1, activation='sigmoid'))

model.summary()
```

```python
model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])
```

```python
model.fit(X_train, y_train, epochs=10, validation_data=(X_test, y_test))
```

```python
model.predict(sequences)
```

```python
model2 = Sequential()
model2.add(Embedding(10000, output_dim=2, input_length=200))
model2.add(SimpleRNN(32, return_sequences=False))
model2.add(Dense(1, activation='sigmoid'))
```

```python
model2.summary()
```

```python
model2.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
history = model2.fit(X_train, y_train, epochs=8, validation_data=(X_test, y_test))
```
